# Central Limit Theorem

The Central Limit Theorem (CLT) states that when we take many independent observations from the same distribution and compute their average, the distribution of the average becomes close to a normal distribution as the sample size increases.


In other terms, suppose $X_1, X_2, \dots, X_n$ are independent and identically distributed random variables with

$$
E[X_i] = \mu, \qquad \mathrm{Var}(X_i) = \sigma^2.
$$

Let the sample mean be

$$
\bar{X}_n = \frac{1}{n}\sum_{i=1}^{n} X_i.
$$

Then, as $n \to \infty$,

$$
\frac{\bar{X}_n - \mu}{\sigma / \sqrt{n}}
$$

approaches a standard normal distribution.

## Background

Let's see the Central Limit Theorem in action. For a fair six-sided die, possible outcomes are {1, 2, 3, 4, 5, 6}.  

The population mean and variance are:

$$
\mu = 3.5, \quad \sigma^2 = \frac{35}{12}, \quad \sigma = \sqrt{\frac{35}{12}}.
$$

If $\bar{X}_n$ represents the average of n independent dice rolls, then:

$$
\mathrm{SD}(\bar{X}_n) = \frac{\sigma}{\sqrt{n}}.
$$

According to the Central Limit Theorem (CLT), as n increases, the distribution of $\bar{X}_n$ approaches a Normal distribution with mean μ and variance σ²/n.

## Question 4: Exploring n and M

Now you get to control the experiment. Use the two sliders below to change:

- **n**: the number of dice rolled in each experiment (each experiment gives one average)
- **M**: the number of times we repeat the experiment (this is how many averages we collect)

The histogram shows all M sample means. The red curve is the Normal curve predicted by theory, and the title compares the predicted spread (SD) with the simulated spread.

Try these, changing **one slider at a time**:
1. Keep M fixed and drag **n** from 1 up to 100. What happens to the shape and spread?
2. Keep n fixed and drag **M** from 10 up to 10,000. What changes, and what stays the same?

```python
import ipywidgets as widgets
from ipywidgets import interact

mu = 3.5
sigma = np.sqrt(35 / 12)

def explore_clt(n, M):
    # Simulate M sample means, each averaging n dice rolls
    means = simulate_means(n, M)

    # Theory: predicted spread of the sample mean
    sdbar = sigma / np.sqrt(n)

    # Bin edges lined up with the possible values of the sample mean
    # (these are multiples of 1/n), so the bars don't look uneven
    step = max(1, int(np.ceil(n / 10))) / n
    bins = np.arange(1 - 1 / (2 * n), 6 + 1 / (2 * n) + step, step)

    fig, ax = plt.subplots(figsize=(8, 4))
    ax.hist(means, bins=bins, density=True, edgecolor="white",
            label="Simulated sample means")

    xs = np.linspace(0.5, 6.5, 600)
    ax.plot(xs, norm.pdf(xs, loc=mu, scale=sdbar), color="red", linewidth=2,
            label="Normal curve predicted by theory")

    ax.set_xlim(0.5, 6.5)   # fixed axis so you can see the spread change
    ax.set_xlabel("Sample mean")
    ax.set_ylabel("Density")
    ax.set_title(f"n = {n}, M = {M}\n"
                 f"Theory SD = {sdbar:.3f}   Simulated SD = {means.std(ddof=1):.3f}")
    ax.legend()
    plt.show()

interact(
    explore_clt,
    n=widgets.IntSlider(value=10, min=1, max=100, step=1,
                        description="n (rolls)", continuous_update=True),
    M=widgets.SelectionSlider(options=[10, 50, 100, 500, 1000, 5000, 10000],
                              value=1000, description="M (repeats)",
                              continuous_update=True),
);
```
