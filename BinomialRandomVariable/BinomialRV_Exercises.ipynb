When we run $n$ independent Bernoulli trials with the same $p$ and where $Y_i$ represents the outcome of the $i\text{th}$ Bernoulli trial, the total number of successes
$$X=\sum_{i=1}^{n} Y_i$$
follows a $Binomial(n, p)$ distribution. The value $Y_i$ will either be 0, a failure, or 1, a success, and each value of $Y_i$ independently follows a Bernoulli($p$) distribution. A good way to think about this is: if I flip $n$ coins, how many times will the coin land on heads?

Using this connection, we will implement binomial simulation by generating 0/1 outcomes and summing them.

The function below generates $n$ independent Bernoulli trials with success probability $p$. Run the cell below to see the output.

```python
def bernoulli(n: int, p: float) -> List:
    # Bernoulli(p): Y = 1[U < p], where U ~ Uniform(0,1)
    U = np.random.rand(n)
    Y = (U < p).astype(int).tolist()
    return Y

bernoulli(10, 0.5)
```

Recall that a binomial random variable represents the **number of successes** $X$ observed in a fixed number of independent trials $n$, where each trial has the same success probability $p$. 

Therefore, the relationship between Bernoulli trials and a binomial random variable is that a binomial random variable is the sum of $n$ Bernoulli trials. 

Here is a function that generates one sample from a distribution $X \sim Binomial(n, p)$.

```python
def binomial_once(n: int, p: float) -> int:
    # Generate n Bernoulli(p) outcomes (0/1)
    trials = bernoulli(n, p) # SOLUTION

    # Compute and return the binomial result.
    return int(np.sum(trials)) # SOLUTION
```

If you are curious, below is an explanation of why the PMF formula works. We will first break down the formula into two parts, the binomial coefficient $\binom{n}{k}$ and $p^k(1-p)^{n-k}$, then analyze their product

The **probability mass function (PMF)** for a **binomial distribution** $X \sim Binomial(n,p)$ is given by the following formula:

$$P_X(k) = \begin{cases}
    \binom{n}{k}  p^k(1-p)^{n-k}  & \text{for } k = 0, 1, 2, 3, \dots, n \\ 
    0 & \text{otherwise} 
\end{cases}
$$

### 1. Binomial Coefficient

Recall that $\binom{n}{k}$ is the **binomial coefficient** and is equal to $\frac{n!}{k!(n-k)!}$. It represents the amount of ways you can pick an unordered set of $k$ items from a set of with $n$ without replacement, commonly shortened to "$n$ choose $k$". 

**Understanding the Numerator**

The only element in the numerator is $n!$. 

$n!$ represents the number of ways $n$ items can be arranged in a unique order without replacement and is equivalent to $n * (n-1) * (n-2) * (n-3) * \dots$

`Example:`

Imagine a container filled with 5 differently colored balls $(n=5)$. 

When the jar is full, the first time you draw a ball, it can be any of 5 colored balls. After the first draw, 4 balls remain, so the second draw has only 4 possibilities. After the second draw, only 3 balls remain so there are 3 possibilities. This continues until only 1 ball remains. The total number of combinations is equal to the number of possibilities at the first draw times the number of possibilities of the second draw times the number of possibilities of the third draw times the number of possibilities of the fourth draw times the number of possibilities of the fifth draw. This is $5 * 4 * 3 * 2 * 1 = 5! = 120$. 

**Understanding the Denominator**

The denominator contains 2 elements: $k!$ and $(n-k)!$

Similarly to the numerator, $k!$ represents the number of ways $k$ items can be arranged in a unique order without replacement and is equivalent to $k * (k-1) * (k-2) * (k-3) * \dots$

Likewise, $(n-k)!$ represents the number of ways $n-k$ items can be arranged in a unique order without replacement and is equivalent to $(n-k) * (n-k-1) * (n-k-2) * \dots$

$(n-k)!$ can also be said to be the number of ways the unselected items can be shuffled.

Multiplied together, it gives the total number of redundant, ordered ways to build the exact same split of items. It counts every way our group with $k$ elements can be arranged and combines them with every way our unwanted group with $n-k$ elements can be arranged.

`Example:`

Imagine a container filled with 5 differently colored balls $(n=5)$. You are concerned about outcomes where the first two draws are some combination of red and blue $(k=2)$ and are not concerned about the last 3 draws. 

In the numerator, consider results where the first two draws are red and blue include $\{red, blue\}$ with $\{yellow, green, orange\}$ left over, $\{red, blue\}$ with $\{orange, yellow, green\}$ left over, and $\{blue, red\}$ with $\{yellow, green, orange\}$ left over, etc. They are counted separately but we consider them all the same: successes. How many duplicate successes are there? The only two ways to order red and blue draws are red first then blue second and the opposite, $2!$. The remaining 3 balls can be arranged in  $3!$ ways. We multiply them to combine them together. This $2! * 3! = (2 * 1) * (3 * 2 * 1) = 12$

**Putting It Together**

The numerator $n!$ shows all ways to order a set of $n$ items.

The denominator $k!(n-k)!$ is the count of all redundant, ordered ways to select $k$ items.

By dividing the numerator by the denominator, we get rid of all the duplicates and are left with the number of unordered ways to pick $k$ items.

`Example:`

Imagine a container filled with 5 differently colored balls $(n=5)$. You are concerned about outcomes where the first two draws are some combination of red and blue $(k=2)$ and are not concerned about the last 3 draws. 

There are $5! = 120$ ways to pick 5 balls. There are $2! * 3! = 12$ duplicate ways to arrange 5 balls where the first 2 picks are from our chosen group. To get the unordered ways to pick 2, we need to count all those 12 ways as 1, and we do this by dividing 120 by 12. $120 / 12 = 10$, meaning there are 10 ways to pick 2 items from 5.

### 2. $p^k(1-p)^{n-k}$

The expression $p^k(1-p)^{n-k}$ calculates the exact probability of getting one specific sequence of $k$ successes and $n-k$ failures in $n$ independent trials.

$p^k$: the probability of having $k$ successes with probability $p$.

$(1-p)^{n-k}$: the probability of $n-k$ failures. If there are $k$ successes out of $n$ trials, there must be $n-k$ failures.

### 3. $\binom{n}{k} p^k(1-p)^{n-k}$

Multiplying the number of ways to choose which $k$ trials are successes by the probability of one specific sequence gives you the total probability of getting $k$ successes in $n$ trials, regardless of the order they happen.

As a reminder, the **probability mass function (PMF)** for a **binomial distribution** $X \sim Binomial(n,p)$ is given by the following formula:

$$P_X(k) = \begin{cases}
    \binom{n}{k}  p^k(1-p)^{n-k}  & \text{for } k = 0, 1, 2, 3, \dots \\ 
    0 & \text{otherwise} 
\end{cases}
$$

This formula only gives the probability that the Bernoulli trials sum up to k, but is not inclusive of outcomes greater than or less than k. For that, we need the cumulative probability.

Remember that for discrete random variables, the cumalative probability can be acquired by summing up outcomes. 

Therefore, the **cumulative distribution function (CDF)** for a **binomial distribution** $X \sim Binomial(n,p)$ is given by the following formulas:

$$P(x \leq k) = \sum_{i=0}^k P_X(i) = \sum_{i=0}^k \binom{n}{i} p^i(1-p)^{(n-i)}$$

To find the probability of getting values greater than equal to k, subtract from 1 the probability that we only get values less than k.
$$P(x \geq k) = 1 - \sum_{i=0}^{k-1} P_X(i) = 1 - \sum_{i=0}^{k-1} \binom{n}{i} p^i(1-p)^{(n-i)}$$

```python
def binom_pmf(k: int, n: int, p: float) -> float:
    # Return P(X = k) for X ~ Binomial(n, p).
    return math.comb(n, k) * (p ** k) * ((1 - p) ** (n - k))

def binom_cdf(k: int, n: int, p: float) -> float:
    # Return P(X <= k) for X ~ Binomial(n, p).
    return sum(binom_pmf(i, n, p) for i in range(k + 1))
```

## Summary

- Binomial random variables is the sum of the result of $n$ Bernoulli trials
- We can create an empirical binomial distribution by sampling binomial random variables from an underlying distribution
-  `np.mean` and `np.var` can be used on a sample distribution to estimate the expected value or variance of a binomial distribution (ie. `np.mean(sample)`, `np.var(sample)`)
- The formula for the theoretical expected value of a binomial distribution is $np$
- The formula for the theoretical expected value of a binomial distribution is $np(1-p)$
- We can estimate the PMF of a binomial distribution by sampling a distribution by taking the number of times each outcome appears and dividing by $n$ (ie. `np.bincount(sample, minLength=n+1) / len(sample`)
- The theoretical formula of the PMF of a binomial distribution is given by $P_X(k) = \begin{cases}
    \binom{n}{k}  p^k(1-p)^{n-k}  & \text{for } k = 0, 1, 2, 3, \dots, n \\ 
    0 & \text{otherwise} 
\end{cases}
$
- The cumulative probability is the probability that the sampled binomial random variable is either less than or greater than a value $k$
- $P(x \leq k) = \sum_{i=0}^k P_X(i) = \sum_{i=0}^k \binom{n}{i} p^i(1-p)^{(n-i)}$
- $P(x \geq k) = 1 - \sum_{i=0}^{k-1} P_X(i) = 1 - \sum_{i=0}^{k-1} \binom{n}{i} p^i(1-p)^{(n-i)}$
- To estimate a cumulative probability, get the number of times each outcome occurs and either sum the probabilities you are interested in or subtract from 1 the probabilities you are not interested in (ie. `sum([sample[i] for i in range (0, k)])` or `1 - sum([sample[i] for i in range(0, k)])`)
