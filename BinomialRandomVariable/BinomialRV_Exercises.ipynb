When we run $n$ independent Bernoulli trials with the same $p$, the total number of successes
$$X=\sum_{i=1}^{n} Y_i$$
follows a $Binomial(n, p)$ distribution. Using this connection, we will implement binomial simulation by generating 0/1 outcomes and summing them, then verify key properties (e.g., mean and variance).

Here is an function for generating N bernoulli samples. Run the following cell see what the output looks like.

```python
def bernoulli(N: int, p: float) -> List:
    # Bernoulli(p): Y = 1[U < p], where U ~ Uniform(0,1)
    U = np.random.rand(N)
    Y = (U < p).astype(int).tolist()
    return Y

bernoulli(10, 0.5)
```

Recall that binomial model describes the **number of successes** $X$ observed in a fixed number of independent trials $n$, where each trial has the same success probability $p$. 

Therefore, the relationship between Bernoulli trials and a binomial random variable is that a binomial random variable is the sum of $n$ Bernoulli trials. 

Here is a function that generates one binomial variable from a distribution $X \sim Binomial(n, p)$

```python
def binomial_once(n: int, p: float) -> int:
    # Generate n Bernoulli(p) outcomes (0/1)
    trials = bernoulli(n, p) # SOLUTION

    # Compute and return the binomial result.
    return int(np.sum(trials)) # SOLUTION
```

Next, we will compare our simulated results to those acquired through the **probability mass function (PMF)** formula for binomial distributions.

As a reminder, the **probability mass function (PMF)** for a **binomial distribution** $X \sim Binomial(n,p)$ is given by the following formula:

$$P_X(k) = \begin{cases}
    \binom{n}{k}  p^k(1-p)^{n-k}  & \text{for } k = 0, 1, 2, 3, \dots \\ 
    0 & \text{otherwise} 
\end{cases}
$$

Recall that $\binom{n}{k}$ is the **binomial coefficient** and is equal to $\frac{n!}{k!(n-k)!}$. It represents the amount of ways you can pick an unordered set of $k$ items from a set of with $n$ without replacement, commonly shortened to "$n$ choose $k$". 

Below, we break down the numerator and denominator of the formula and use an example involving colored balls to explain why this formula works.

**Understanding the Numerator**

The only element in the numerator is $n!$. 

$n!$ represents the number of ways $n$ items can be arranged in a unique order without replacement and is equivalent to $n * (n-1) * (n-2) * (n-3) * \dots$

`Example:`

Imagine a container filled with 5 differently colored balls $(n=5)$. 

When the jar is full, the first time you draw a ball, it can be any of 5 colored balls. After the first draw, 4 balls remain, so the second draw has only 4 possibilities. After the second draw, only 3 ball remains so there are 3 possibilities. This continues until only 1 ball remains. The total number of combinations is equal to the number of possibilities at the first draw times the number of possibilities of the second draw times the number of possibilities of the third draw times the number of possibilities of the fourth draw times the number of possibilities of the fifth draw. This is $5 * 4 * 3 * 2 * 1 = 5! = 120$. 

**Understanding the Denominator**

The denominator contains 2 elements: $k!$ and $(n-k)!$

Similarly to the numerator, $k!$ represents the number of ways $k$ items can be arranged in a unique order without replacement and is equivalent to $k * (k-1) * (k-2) * (k-3) * \dots$

Likewise, $(n-k)!$ represents the number of ways $n-k$ items can be arranged in a unique order without replacement and is equivalent to $(n-k) * (n-k-1) * (n-k-2) * \dots$

$(n-k)!$ can also be said to be the number of ways the unselected items can be shuffled.

Multiplied together, it gives the total number of redundant, ordered ways to build the exact same split of items. It counts every way our group with $k$ elements can be arranged and combines them with every way our unwanted group with $n-k$ elements can be arranged.

`Example:`

Imagine a container filled with 5 differently colored balls $(n=5)$. You are concerned about outcomes where the first two draws are some combination of red and blue $(k=2)$ and are not concerned about the last 3 draws. 

In the numerator, consider results where the first two draws are red and blue include $\{red, blue\}$ with $\{yellow, green, orange\}$ left over, $\{red, blue\}$ with $\{orange, yellow, green\}$ left over, and $\{blue, red\}$ with $\{yellow, green, orange\}$ left over, etc. They are counted separately but we consider them all the same: successes. How many duplicate successes are there? The only two ways to order red and blue draws are red first then blue second and the opposite, $2!$. The remaining 3 balls can be arranged in  $3!$ ways. We multiply them to combine them together. This $2! * 3! = (2 * 1) * (3 * 2 * 1) = 12$


**Putting It Together**

The numerator $n!$ shows all ways to order a set of $n$ items.

The denominator $k!(n-k)!$ the count of all redundant, ordered ways to select $k$ items.

By dividing the numerator by the denominator, we get rid of all the duplicates and are left with the number of unordered ways to pick $k$ items.

`Example:`

Imagine a container filled with 5 differently colored balls $(n=5)$. You are concerned about outcomes where the first two draws are some combination of red and blue $(k=2)$ and are not concerned about the last 3 draws. 

There are $5! = 120$ ways to pick 5 balls. There are $2! * 3! = 12$ duplicate ways to arrange 5 balls where the first 2 picks are from our chosen group. To get the unordered ways to pick 2, we need to count all those 12 ways as 1, and we do this by dividing 120 by 12. $120 / 12 = 10$, meaning there are 10 ways to pick 2 items from 5.

The function below calculates the binomial coefficient for you.

```python
def binom(n, k):
    numerator = math.factorial(n)
    denominator = math.factorial(k) * math.factorial(n-k)

    return numerator / denominator
```

## Part 3 — Using Binomial Random Variables To Solve Probability Problems

This part explores one example of a real-world application where you can use binomial distributions and random variables to find likelihoods. 

### Scenario: Pokemon Shiny Hunting

You just caught your first shiny Pokemon! Now, you are obsessed with having a full team of shiny Pokemon to battle the Elite 4 with. With this new goal in mind, you are curious about the probability of getting 5 more shiny Pokemon naturally from just progressing the game.

Assuming you will run into 15000 more encounters in your game of Pokemon, what are the chances you run into at least 5 more shiny Pokemon when the chance of encountering a shiny is 1 in 4096.

As a reminder, the **probability mass function (PMF)** for a **binomial distribution** $X \sim Binomial(n,p)$ is given by the following formula:

$$P_X(k) = \begin{cases}
    \binom{n}{k}  p^k(1-p)^{n-k}  & \text{for } k = 0, 1, 2, 3, \dots \\ 
    0 & \text{otherwise} 
\end{cases}
$$

This formula only gives the probability that the Bernoulli trials sum up to k, but is not inclusive of outcomes greater than or less than k. For that, we need the cumulative probability.

Remember that for discrete random variables, the cumalative probability can be acquired by summing up outcomes. 

Therefore, the **cumulative distribution function (CDF)** for a **binomial distribution** $X \sim Binomial(n,p)$ is given by the following formulas:

$$P(x \leq k) = \sum_{i=0}^k P_X(i) = \sum_{i=0}^k \binom{n}{i} p^i(1-p)^{(n-i)}$$

To find the probability of getting values greater than equal to k, subtract from 1 the probability that we only get values less than k.
$$P(x \geq k) = 1 - \sum_{i=0}^{k-1} P_X(i) = 1 - \sum_{i=0}^{k-1} \binom{n}{i} p^i(1-p)^{(n-i)}$$

```python
def binom_pmf(k: int, n: int, p: float) -> float:
    # Return P(X = k) for X ~ Binomial(n, p).
    return math.comb(n, k) * (p ** k) * ((1 - p) ** (n - k))

def binom_cdf(k: int, n: int, p: float) -> float:
    # Return P(X <= k) for X ~ Binomial(n, p).
    return sum(binom_pmf(i, n, p) for i in range(k + 1))
```
