# The St. Petersburg Paradox

The **St. Petersburg Paradox** is a probability game involving repeated flips of a fair coin. Although the rules are simple, the game produces a surprising result when we examine its potential payouts.

---

### How Does the Game Work?

The game follows three simple rules:

1. A fair coin is flipped repeatedly until the first **heads (H)** appears.
2. The payout starts at **$2** and doubles for each additional flip needed.
3. The game ends when **heads (H)** appears, and the player receives the payout.

### Understanding the Payouts

The payout depends on how many flips occur before the first heads appears.



| Coin Flip Sequence | Number of Flips | Payout |
|:-------------------|:---------------:|-------:|
| H | 1 | &#36;2 |
| TH | 2 | &#36;4 |
| TTH | 3 | &#36;8 |
| TTTH | 4 | &#36;16 |
| TTTTH | 5 | &#36;32 |



Here, **(H)** represents heads and **(T)** represents tails. The payout **doubles with each additional flip**, while the probability of obtaining a larger payout decreases.

### The Payout Formula

If the first heads appears on flip $n$, the payout is given by:

$$
X = 2^n
$$

where:

- $X$ represents the payout in dollars.
- $n$ represents the number of flips until the first heads appears.

---

Although the game offers the possibility of extremely large payouts, these outcomes become increasingly unlikely.

## Exercise 2: Expected Payout and the Paradox


### Understanding Expected Value

In Exercise 1, we observed that most games produce small payouts, while exceptionally large payouts occur infrequently. This raises an important question:


<div style="width: 100%; text-align: center; margin: 20px 0;">
    <strong>How can we determine what a game is worth when its possible payouts vary so greatly?</strong>
</div>

To investigate this, let's introduce **expected value**, which accounts for both the size of each possible payout and its probability. 
- Each outcome contributes to the expected value according to:

$$
\text{Contribution} = \text{Payout} \times \text{Probability}
$$

### Expected Payout in the St. Petersburg Game

The following table illustrates how the first three possible outcomes contribute to the expected payout.

| First Heads Appears | Payout | Probability | Contribution |
|:-------------------|-------:|:-----------:|-------------:|
| Flip 1 | &#36;2 | $1/2$ | &#36;1 |
| Flip 2 | &#36;4 | $1/4$ | &#36;1 |
| Flip 3 | &#36;8 | $1/8$ | &#36;1 |

Notice that although the payout doubles with each additional flip, **each outcome contributes the same amount to the expected value**.

More generally, if the first heads appears on flip $n$, the payout is $2^n$ and its probability is $(1/2)^n$.

Adding the contributions from every possible outcome gives:

$$
E[X] = \sum_{n=1}^{\infty} 2^n \left(\frac{1}{2}\right)^n
$$

Here, $E[X]$ represents the expected payout before subtracting any entry fee.

## What's the Paradox?

In Exercise 1, we observed that most games produce small payouts, while exceptionally large payouts are rare. Exercise 2 revealed a surprising result: **the theoretical expected payout of the St. Petersburg game is infinite**.

<div style="width: 100%; text-align: center; margin: 20px 0;">
    <strong>If the expected payout is infinite, why are people unwilling to pay extremely high entry fees?</strong>
</div>

This is the **St. Petersburg Paradox**. Although the mathematical expected payout grows without bound, people are generally unwilling to pay enormous amounts to participate.

### Understanding Diminishing Marginal Utility

One explanation comes from the economic idea of diminishing marginal utility. Utility is the value or satisfaction someone receives from an outcome. If a person gets more and more money, each additional dollar may provide less additional satisfaction. Receiving an extra 100 dollars might make a big difference to someone with 200 dollars, but much less difference to someone with 100,000 dollars.
