## The St. Petersburg Paradox

The St. Petersburg Paradox is a game that involves flipping a fair coin until the first head appears. The initial payout begins at $2 and doubles each time tails is flipped. Once heads is flipped, the game ends and the player receives the current payout.

The table below shows how the payout changes depending on when the first head appears:


| Coin Flip Sequence | Payout |
|:-------------------|-------:|
| H                  | &#36;2  |
| TH                 | &#36;4  |
| TTH                | &#36;8  |
| TTTH               | &#36;16 |

Each flip is independent & heads and tails are equally likely. If the first head appears on flip (n), the payout in dollars is: 
$$
X = 2^n
$$


A large payout is possible, but how likely is it? Let's find out!

```python
# DO NOT MODIFY
def games_until_large_profit():
    entry_fee = 10
    games_played = 0

    while True:
        games_played += 1
        net_gain = play()[2] - entry_fee

        if net_gain >= 100:
            return games_played


waiting_times = [games_until_large_profit() for _ in range(1000)]

plt.figure(figsize=(10, 4))
plt.hist(waiting_times, bins=30, color="skyblue", edgecolor="white")
plt.axvline(
    np.mean(waiting_times),
    color="red",
    linestyle="--",
    label=f"Average waiting time: {np.mean(waiting_times):.1f} games"
)
plt.xlabel("Number of games until the first profit of at least $100")
plt.ylabel("Number of experiments")
plt.title("How Long Is the Wait for a Large Profit?")
plt.legend();
```

To recap, exercise 1 showed that most games produce small payouts, while a few rare games produce much larger ones. With this in mind, how much weight should those rare payouts receive when deciding what is reasonable to pay?

The **expected value** combines the size of each payout with its probability. Each payout contributes its amount multiplied by its probability:

$$
\text{Contribution} = \text{Payout} \times \text{Probability}.
$$

For example, a $\$2$ payout occurs with probability $1/2$, so its contribution is $\$2 \times 1/2 = \$1$.

| First heads appears on | Payout | Probability | Contribution to expected payout |
|---|---|---|---|
| Flip $1$ | $\$2$ | $1/2$ | $\$1$ |
| Flip $2$ | $\$4$ | $1/4$ | $\$1$ |
| Flip $3$ | $\$8$ | $1/8$ | $\$1$ |

Adding the contributions from every possible outcome gives the expected payout:

$$
E[X] = \sum_{n=1}^{\infty} 2^n \left(\frac{1}{2}\right)^n,
$$

where $n$ is the flip on which the first heads appears. This calculation describes one game's possible payouts before having to subtract an entry fee.

There are infinitely many possible outcomes to St. Petersburg, so the function that will be below this will begin by adding a finite number of contributions. The input `max_flips` specifies the latest first-heads outcome included. For example, `max_flips = 3` includes the three rows above.

Games lasting longer than `max_flips` remain possible, but their contributions will not be included in this partial sum. Later on, we will explore what happens as more outcomes are included.

## What's the Paradox?

In Exercise 1, we saw that most games produce small payouts, while very large payouts are possible but rare. In Exercise 2 so far, we discovered something surprising: the expected payout of the game is infinite!

If the expected payout is infinite, why wouldn't someone be willing to pay $100, $1,000, or even more to play?

This is the **St. Petersburg paradox**. The mathematical expected payout grows without bound, yet people usually don't want to pay enormous amounts to play the game. How can we explain this difference?

One explanation comes from the economic idea of diminishing marginal utility. Utility is the value or satisfaction someone receives from an outcome. If a person gets more and more money, each additional dollar may provide less additional satisfaction. For example, receiving an extra `$100` might make a big difference to someone with `$200`, but much less difference to someone with `$100,000`.

Let's stick with this idea of utility! Let's now consider the table below. It compares the payouts in the St. Petersburg game with a rough measure of their utility.


| Payout | Payout doubles? | Rough utility |
|:---|:---|---:|
| \$2 | — | 0.69 |
| \$4 | Yes | 1.39 |
| \$8 | Yes | 2.08 |
| \$16 | Yes | 2.77 |
| \$32 | Yes | 3.47 |
| \$64 | Yes | 4.16 |
| \$128 | Yes | 4.85 |


What's important here is that while the payout doubles each time, the increase in utility is about the same amount. Utility increases much more slowly than the payout.
