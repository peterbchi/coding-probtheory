## The St. Petersburg Paradox

The St. Petersburg Paradox is a game that involves flipping a fair coin until the first head appears. The initial payout begins at $2 and doubles each time tails is flipped. Once heads is flipped, the game ends and the player receives the current payout.

The table below shows how the payout changes depending on when the first head appears:


| Coin Flip Sequence | Payout |
|:-------------------|-------:|
| H                  | &#36;2  |
| TH                 | &#36;4  |
| TTH                | &#36;8  |
| TTTH               | &#36;16 |

Each flip is independent & heads and tails are equally likely. If the first head appears on flip (n), the payout in dollars is: 
$$
X = 2^n
$$


A large payout is possible, but how likely is it? Let's find out!

**Question 1.5.** So, how long does it take to see a large profit?

In the previous question, it was shown to us that profits of at least $\$100$ occur rarely. Let's now consider how many games might be played before one like this occurs!

Continuing with a $\$10$ entry fee, a net gain of at least $\$100$ requires a payout of at least $\$128$ because the possible payouts are powers of $2$.

The function that is provided to generate this code reuses `play()` from **Question 1.1** to count the number of games until the first net gain of at least $\$100$. It repeats this experiment 1000 times and graphs the number of games it takes to see a profit like this.

Run the cell below!

```python
# DO NOT MODIFY
def games_until_large_profit():
    entry_fee = 10
    games_played = 0

    while True:
        games_played += 1
        net_gain = play()[2] - entry_fee

        if net_gain >= 100:
            return games_played


waiting_times = [games_until_large_profit() for _ in range(1000)]

plt.figure(figsize=(10, 4))
plt.hist(waiting_times, bins=30, color="skyblue", edgecolor="white")
plt.axvline(
    np.mean(waiting_times),
    color="red",
    linestyle="--",
    label=f"Average waiting time: {np.mean(waiting_times):.1f} games"
)
plt.xlabel("Number of games until the first profit of at least $100")
plt.ylabel("Number of experiments")
plt.title("How Long Is the Wait for a Large Profit?")
plt.legend();
```

A payout of at least $\$100$ requires the first six flips to be tails (5 tails flips is $\$32$ & 6 tails flips is $\$128$). The probability in each game is $(1/2)^6 = 1/64$, giving an average of $64$ games before you theoretically see the first large payout.

## What's the Paradox?
This notion is what makes up the St. Petersburg paradox - As the number of plays increases, the expected value increases without bound! If an infinite money is expected, why don't people pay more to play?

We can explain this with the Law of Diminishing Marginal Utility: **as consumption increases, the increase in consequent satisfaction decreases.**

The utility in this scenario refers to the perceived value or satisfaction from winning money. As the odds of winning decrease with each subsequent play, people weigh the both the risks and the potential degree of satisfaction during their decision-making process.  

Mathematician Daniel Bernoulli proposed a logarithmic utility function to represent the expected utility of playing $U(w) = l n(w)$. Under a logarithmic model, EV becomes finite rather than infinite and the paradox disappears because the assumption that there is constant and infinite financial gain is no longer feasible. Expected value can yield extreme results that do not reflect external factors!
