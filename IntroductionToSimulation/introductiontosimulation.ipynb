### **2.2. Analyzing Numerical Data**

When analyzing simulations, we often want to **summarize our results and understand how much they vary**.

NumPy provides several useful functions to help us accomplish this!

| Function | Description |
|:---|:---|
| `np.sum()` | Calculates the sum of all values. |
| `np.mean()` | Calculates the average of the values. |
| `np.median()` | Finds the middle value when the data is sorted. |
| `np.var()` | Measures how spread out the values are by calculating their variance. |
| `np.std()` | Calculates the standard deviation, describing the spread around the mean. |
| `np.sqrt()` | Calculates the square root of a number. |


<span style="display: inline-block; background-color: #fff1f2; color: #9f1239; border-left: 4px solid #fb7185; padding: 10px 16px; border-radius: 5px; font-size: 17px; font-weight: bold;">💡 Let's Look at an Example!</span>


Imagine we recorded the results of rolling a fair six-sided die **10 times**.

- These functions can be used to summarize the outcomes and understand how much they vary.

```python
# DO NOT MODIFY.

# Create an array containing the outcomes of rolling a six-sided die 10 times.
rolls = np.array([1, 2, 2, 3, 4, 4, 4, 6, 6, 6])

# Use NumPy functions to calculate the sum, mean, median, variance, and standard deviation of the die rolls.
total = np.sum(rolls)
mean = np.mean(rolls)
median = np.median(rolls)
variance = np.var(rolls)
standard_deviation = np.std(rolls)

# Use np.sqrt() to verify that the standard deviation is the square root of the variance.
square_root_of_variance = np.sqrt(np.var(rolls))

# Display results in a table
results_table = pd.DataFrame({
    "Statistic": [
        "Sum",
        "Mean",
        "Median",
        "Variance",
        "Standard deviation",
        "Square root of variance"
    ],
    "Value": [
        total,
        mean,
        median,
        variance,
        standard_deviation,
        square_root_of_variance
    ]
})

display(results_table)
```

You should have noticed that these functions describe **different characteristics of the same data**.

- **Mean and median:** Help describe the center of our results.
- **Variance and standard deviation:** Help describe how spread out our results are.

You should also have noticed that `np.std(rolls)` and `np.sqrt(np.var(rolls))` produced the same result!

### **2.3. Counting Outcomes**

Another important part of simulation is determining **how frequently certain outcomes occur**.

NumPy provides several useful functions for counting outcomes, such as:

| Function | Description |
|:---|:---|
| `np.unique()` | Identifies distinct values in an array and can count their occurrences using `return_counts=True`. |
| `np.bincount()` | Counts how often each nonnegative integer appears in an array. |
| `np.count_nonzero()` | Counts nonzero values or outcomes that satisfy a condition. |


<span style="display: inline-block; background-color: #fff1f2; color: #9f1239; border-left: 4px solid #fb7185; padding: 10px 16px; border-radius: 5px; font-size: 17px; font-weight: bold;">💡 Let's Look at an Example!</span>


Using our previous die-roll results, let's determine **how often each outcome appeared** and how many rolls were greater than 4.


```python
# DO NOT MODIFY.

# Create an array containing the outcomes of rolling a six-sided die 10 times.
rolls = np.array([1, 2, 2, 3, 4, 4, 4, 6, 6, 6])

# Use np.unique() to identify the unique outcomes and count how often each one appears.
values, counts = np.unique(rolls, return_counts=True)

# Use np.bincount() to count how many times each integer from 0 through 6 appears.
frequency = np.bincount(rolls, minlength=7)

# Use np.count_nonzero() to count how many rolls resulted in a number greater than 4.
rolls_above_four = np.count_nonzero(rolls > 4)

# Display the calculated counts and frequencies in a table for comparison.
results_table = pd.DataFrame({
    "Statistic": [
        "Unique outcomes",
        "Counts",
        "Frequency of outcomes 1 through 6",
        "Rolls greater than 4"
    ],
    "Value": [
        str(values),
        str(counts),
        str(frequency[1:]),
        rolls_above_four
    ]
})

display(results_table)
```

Both `np.unique()` and `np.bincount()` help us **count how frequently different outcomes occur**, while `np.count_nonzero()` counts outcomes that satisfy a specific condition.

Together, these functions allow us to **analyze outcome frequencies and estimate probabilities** in our simulations.

### **2.4. Rounding Numerical Results**

Simulation results often contain **decimal numbers with many digits**, making them difficult to read.

The `np.round()` function allows us to **round numerical values to a specified number of decimal places**, making our results easier to interpret.

For example, rounding **3.14159** to **2 decimal places**:

- `np.round(3.14159, 2)` → **3.14**


<span style="display: inline-block; background-color: #fff1f2; color: #9f1239; border-left: 4px solid #fb7185; padding: 10px 16px; border-radius: 5px; font-size: 17px; font-weight: bold;">💡 Let's Look at an Example!</span>


Let's use `np.round()` to display the mean and standard deviation of our die-roll results to **2 decimal places**. Run the cell below!

```python
# DO NOT MODIFY.

# Create an array containing the outcomes of rolling a six-sided die 10 times.
rolls = np.array([1, 2, 2, 3, 4, 4, 4, 6, 6, 6])

# Use NumPy functions to calculate the mean and standard deviation of the die rolls.
average = np.mean(rolls)
standard_deviation = np.std(rolls)

# Use np.round() to round both calculated statistics to two decimal places.
rounded_average = np.round(average, 2)
rounded_std = np.round(standard_deviation, 2)

# Display the rounded mean and standard deviation in a table.
results_table = pd.DataFrame({
    "Statistic": ["Average", "Standard deviation"],
    "Value": [rounded_average, rounded_std]
})

display(results_table)
```

### **2.5. Making Simulations Reproducible**

Random simulations typically produce **different outcomes each time they run**. However, sometimes we want to reproduce the same results, especially when testing or comparing simulations.

NumPy provides the **`np.random.seed()`** function to help us accomplish this!

- By setting a seed, we establish a starting point for NumPy's random number generator, allowing us to **reproduce the same sequence of random outcomes**.



<span style="display: inline-block; background-color: #fff1f2; color: #9f1239; border-left: 4px solid #fb7185; padding: 10px 16px; border-radius: 5px; font-size: 17px; font-weight: bold;">💡 Let's Look at an Example!</span>



Let's use `np.random.seed(42)` to generate the same sequence of random die rolls twice.

- **Note**: We'll use `np.random.randint()`, which we'll explore in more detail later in this notebook. As of right now, you don't need to worry what this does!


```python
# DO NOT MODIFY.

# Set the random seed to 42 to make the results reproducible.
np.random.seed(42)

# Generate five random die rolls using np.random.randint().
first_run = np.random.randint(1, 7, size=5)

# Reset the random seed to the same value to reproduce the sequence.
np.random.seed(42)

# Generate another five random die rolls to compare with the first run.
second_run = np.random.randint(1, 7, size=5)

# Display the results of both simulations in a table for comparison.
results_table = pd.DataFrame({
    "Simulation": ["First Run", "Second Run"],
    "Die Rolls": [str(first_run), str(second_run)]
})

display(results_table)
```

You should have noticed that **both simulations generated the exact same outcomes**!

By resetting the seed to the same value, we can reproduce our random results.

As of right now, we've learned how to use `np.random.choice()` to generate a single random outcome and multiple random outcomes using the `size` parameter.

Now, what if we **don't want the same outcome to be selected more than once?**

This is where the `replace` parameter comes in! Recall that `np.random.choice()` has a parameter called `replace`, which determines whether an item can be selected more than once.

There are two ways we can sample:

- **With Replacement (`replace=True`):** After selecting an item, we put it back before making another selection. This means the same item can be selected multiple times.
- **Without Replacement (`replace=False`):** After selecting an item, we do not put it back. This means the same item cannot be selected again.

By design, NumPy sets `np.random.choice()` behavior to replace=True, meaning that sampling is performed with replacement unless specified otherwise.

## **4. Introduction to `np.random.randint()`**


Let's explore another useful NumPy function: **`np.random.randint()`**.

### `np.random.randint()`

The `np.random.randint()` function generates **random integers (whole numbers)** within a specified range. Unlike `np.random.choice()`, it generates random numbers **without listing every possible value**.

### Understanding the Syntax

Let's take a look at the main parameters:

`np.random.randint(low, high=None, size=None)`

| Parameter | Description | Default Value |
|:---|:---|:---|
| `low` | The smallest possible integer (**included**) when `high` is provided. | Required |
| `high` | The upper limit (**not included**). | `None` |

**Important Things to Remember:**

For example, `np.random.randint(1, 7)` generates integers from **1 to 6**, while `np.random.randint(6)` generates integers from **0 to 5**.

### **7.3. Repeating Until a Condition Is Met**

Currently, we learned that `for` loops allow us to repeat an experiment a **specified number of times**. However, some experiments require us to keep repeating until a particular outcome occurs.

This is where **`while` loops** come in!

| Python Tool | Purpose |
|:---|:---|
| `while` | Repeats code as long as a condition remains true. |
| `while True` | Continuously repeats code until the loop is stopped. |
| `break` | Immediately stops the loop when a condition is met. |

---



<span style="display: inline-block; background-color: #fff1f2; color: #9f1239; border-left: 4px solid #fb7185; padding: 10px 16px; border-radius: 5px; font-size: 17px; font-weight: bold;">💡 Let's Look at an Example!</span>

Imagine rolling a fair six-sided die repeatedly **until we roll a 6**.

We don't know how many rolls it will take, so we'll use:

- **`while`** to keep rolling the die.
- **`break`** to stop the simulation once we roll a 6.

Let's see how many rolls it takes!


```python
# DO NOT MODIFY.

# Initialize a counter to track the total number of die rolls.
roll_count = 0

# Use a while loop to repeatedly roll a fair six-sided die until a 6 appears.
while True:
    # Generate a random die roll and increase the counter by 1.
    roll = np.random.randint(1, 7)
    roll_count += 1

    # Stop the loop when the die lands on 6.
    if roll == 6:
        break

# Display the final die roll and the total number of rolls needed to get a 6.
print("Final roll:", roll)
print("Number of rolls needed:", roll_count)
```

You should have noticed that the final roll was always **6**, but the number of attempts could vary each time!

Here's how the code works:

- **`while True:`** keeps repeating the simulation.
- **`roll_count += 1`** increases the counter by 1 after each roll.
- **`if roll == 6:`** checks whether we rolled a 6.
- **`break`** stops the loop once the condition is met.


Remember that `while True` will continue running indefinitely unless a stopping condition, such as `break`, is included!

## **9. Repeating Simulations and Collecting Results**

We've learned how to generate random outcomes using loops and functions. Now, let's take things one step further by **repeating an entire experiment multiple times** and collecting the results!

<span style="display: inline-block; background-color: #fff1f2; color: #9f1239; border-left: 4px solid #fb7185; padding: 10px 16px; border-radius: 5px; font-size: 17px; font-weight: bold;">💡 Let's Look at an Example!</span>

Imagine rolling a fair six-sided die **10 times** and calculating the average outcome.

| Experiment | What Happens? |
|:---|:---|
| **One experiment** | Roll the die 10 times and calculate one average. |
| **100 experiments** | Repeat the 10-roll experiment 100 times and collect 100 averages. |

By collecting these averages, we can explore **how simulation results vary across repeated experiments**.

In this section, we'll use `for` loops, lists, and NumPy to **repeat simulations, store their results, and analyze the collected data**.

```python
# DO NOT MODIFY.

# Create an empty list to store the average outcome from each experiment.
sample_means = []

# Repeat the experiment five times.
for i in range(5):

    # Simulate rolling a fair six-sided die 10 times.
    rolls = np.random.randint(1, 7, size=10)

    # Calculate the average of the 10 rolls and store it in the list.
    sample_means.append(np.mean(rolls))

# Display the average outcome from each experiment in a table.
results_table = pd.DataFrame({
    "Experiment": range(1, 6),
    "Average Die Roll": sample_means
})

display(results_table.round(2))
```


---

<span style="display: inline-block; background-color: #ecfdf5; color: #065f46; border-left: 4px solid #10b981; padding: 10px 16px; border-radius: 5px; font-size: 17px; font-weight: bold;">🔍 What Did We Observe?</span>

You should have noticed that **each experiment produced a different average**, even though we simulated rolling the same fair die 10 times.

Here's how our simulation works:

- **`for i in range(5):`** repeats the entire experiment five times.
- **`np.random.randint(1, 7, size=10)`** generates 10 random die rolls for each experiment.
- **`np.mean(rolls)`** calculates the average outcome of each experiment.
- **`sample_means.append()`** stores each average so we can analyze the results later.

Rather than recording every individual die roll, we're now collecting **one summary statistic from each experiment**!
