## 2. Getting Started with `np.random.choice()`

To start, the first function that will be explored is: **`np.random.choice()`**!

### `np.random.choice()`

The `np.random.choice()` function allows us to randomly select an outcome from a given collection of possible values. This is particularly useful when simulating random experiments, such as flipping coins, rolling dice, or drawing cards.

### Understanding the Syntax

Before we begin using this function, let's take a look at its syntax:

`np.random.choice(a, size=None, replace=True, p=None)`


At first glance, this might seem a little confusing. However, each parameter serves a specific purpose.

| Parameter | Description | Default Value |
|:---|:---|:---|
| `a` | The possible values or outcomes that we want to randomly select from. | Required |
| `size` | The number of random outcomes we want to generate. | `None` (one outcome) |
| `replace` | Determines whether the same outcome can be selected more than once. | `True` |
| `p` | Specifies the probability of selecting each possible outcome. | `None` (equal probabilities) |

**Don't worry if these parameters seem unfamiliar right now!** Throughout this section, we will explore how they work through examples and hands-on coding exercises.

### **Exercise 2.3: Sampling With and Without Replacement**

As of right now, we've learned how to use `np.random.choice()` to generate a single random outcome and multiple random outcomes using the `size` parameter.

Now, what if we **don't want the same outcome to be selected mroe than once?**

This is where the `replace` parameter comes in! Recall that `np.random.choice()` has a parameter called `replace`, which determines whether an item can be selected more than once.

There are two ways we can sample:

- **With Replacement (`replace=True`):** After selecting an item, we put it back before making another selection. This means the same item can be selected multiple times.
- **Without Replacement (`replace=False`):** After selecting an item, we do not put it back. This means the same item cannot be selected again.

By design, NumPy sets `np.random.choice()` behavior to replace=True, meaning that sampling is performed with replacement unless specified otherwise.

#### **Let's Look at an Example!**

Imagine a bag containing four M & M's, each with a different color:


- <span style="color: #FF5252;">Red</span>
- <span style="color: #42A5F5;">Blue</span>
- <span style="color: #66BB6A;">Green</span>
- <span style="color: #FBC02D;">Yellow</span>


Suppose we want to randomly draw **3 M & M's** from this bag. There are two ways we can draw M&Ms from the bag:

- **With Replacement (`replace=True`):** After drawing an M&M, we **put it back** into the bag before the next draw. This means the same M&M can be selected multiple times.

- **Without Replacement (`replace=False`):** After drawing an M&M, we **keep it out** of the bag. This means that the same M&M cannot be selected again.

**Question 2.3:** Using `np.random.choice()`, simulate drawing **3 M & M's** from the bag in two different scenarios.

### **Exercise 2.4: Simulating Unequal Probabilities**

Everything that we've been doing so far had an **equal probability** of occurring.

However, what if some otucomes are **more likely to occur than others?** This is where the **`p` parameter** comes in!

The `p` parameter allows us to assign different probabilities to each possible outcome in `np.random.choice()`.

#### Let's Look at an Example!

Suppose we have a biased coin where:

- **Heads:** 70% probability
- **Tails:** 30% probability

We can simulate flipping this coin using:

```python
np.random.choice(["Heads", "Tails"], p=[0.7, 0.3])
```

Notice how the probabilities in `p` correspond to the outcomes in the same order.

- `0.7` corresponds to **Heads**.
- `0.3` corresponds to **Tails**.

**Important:** The probabilities must add up to **1 (100%)**.

### **2.5. Supplementary Notes**

Now that we've learned how to generate random outcomes using `np.random.choice()`, let's explore a few useful functions for **analyzing simulation results**!

| Function | Description |
|:---|:---|
| `np.sum()` | Calculates the sum of values or counts outcomes that satisfy a condition. |
| `np.mean()` | Calculates the average or proportion of outcomes satisfying a condition. |
| `len()` | Returns the number of elements in a collection. |
| `np.unique()` | Identifies unique values and can count their occurrences using `return_counts=True`. |

### Let's Look at an Example!

Suppose we simulated rolling a fair six-sided die **10 times**. These functions are useful to calculate the total, count specific outcomes, find the average, and analyze how frequently each number appears.

````python
# Simulate rolling a fair six-sided die 10 times & print their results
rolls = np.random.choice([1, 2, 3, 4, 5, 6], size=10)
print("Simulated rolls:", rolls)

# Sum of all die rolls
print("Sum:", np.sum(rolls))

# Number of times we rolled a six
print("Number of sixes:", np.sum(rolls == 6))

# Average die roll
print("Average:", np.mean(rolls))

# Total number of rolls
print("Total rolls:", len(rolls))

# Proportion of rolls that were sixes
print("Proportion of sixes:", np.mean(rolls == 6))

# Unique outcomes and their counts
values, counts = np.unique(rolls, return_counts=True)

print("Unique values:", values)
print("Counts:", counts)
````
