## The Poisson Distribution

As we've covered in the lesson notes, the Poisson Distribution is a discrete probability distribution with probability mass function

$$P(X = k) = \frac{\lambda^ke^{-\lambda}}{k!}$$

where $\lambda$ represents the average rate or expected number of events occurring within a fixed interval.

With the fluctuating rate case, modeling the data using a Poisson distribution **vastly underestimates the variance** because it assumes that the rate stays constant. This is a concept known as <i>overdispersion</i>, where the real-world data has more variation than a statistical model expects.

  Therefore, in order for a Poisson distribution to accurately model the data, the events must follow a constant average rate.
