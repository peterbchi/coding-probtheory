## The Poisson Distribution

The **Poisson distribution** is a discrete probability distribution that models the probability of a certain count of events occurring within a fixed window of time or space. Its **probability mass function (PMF)** gives the probability of seeing exactly $k$ events in the interval:

$$P(X = k) = \frac{\lambda^k e^{-\lambda}}{k!}$$

Here is what each part of the formula means:

- $X$ is the random variable: the number of events that occur in the interval.
- $k$ is a specific number of events we want the probability for ($k = 0, 1, 2, \dots$).
- $\lambda$ (lambda) is the **average rate**: the expected number of events per interval. It is the only parameter of the distribution, and it must be positive.
- $e \approx 2.718$ is Euler's number, the base of the natural logarithm.
- $k!$ is "$k$ factorial", the product $k \times (k-1) \times \dots \times 1$, with $0! = 1$.

**Example.** Suppose a coffee shop averages $\lambda = 2$ customers per minute. The probability that exactly $k = 3$ customers arrive in a given minute is

$$P(X = 3) = \frac{2^3 e^{-2}}{3!} = \frac{8 \cdot e^{-2}}{6} \approx 0.180$$

So there is about an 18% chance of seeing exactly 3 customers in a minute.

## Part 4: Deep Dive into Poisson Conditions

The Poisson distribution is only a good model when the events being counted behave in a certain way. Three conditions matter:

1. **Independence:** One event doesn't make another more or less likely.
2. **Constant average rate:** The average number of events per interval stays the same throughout.
3. **Events happen one at a time:** Two events can't occur at exactly the same moment.

When these conditions hold, the mean and variance of the counts are both $\lambda$, as you saw in Part 3. That gives us a way to check: if the variance is very different from the mean, a Poisson model is probably a poor fit.

## Part 5: Approximating the Binomial with the Poisson

Recall that the Binomial distribution counts the number of successes in a fixed number of independent trials, where each trial has the same chance of success. It has two parameters:

- $n$: the number of trials
- $p$: the probability of success on each trial

For example, if you flip a fair coin $n = 10$ times, the number of observed heads follows a Binomial distribution with $p = 0.5$, because the probability of getting heads for one trial is 0.5.

The Poisson distribution can approximate the Binomial distribution under specific conditions: a large number of trials ($n$) and a small probability of success ($p$). In that case, the Poisson parameter is given by $\lambda = n \times p$.

To help explore this relationship, we've provided a function called `plot_comparison(n, p)`. **Run the cell below, but don't change it.**

Given a number of trials `n` and a success probability `p`, the function:

1. Generates 1,000 random values from a **Binomial($n$, $p$)** distribution.
2. Generates 1,000 random values from a **Poisson($\lambda$)** distribution, where $\lambda = n \times p$.
3. Plots a histogram of both on the same graph: **green** for the Binomial values and **red** for the Poisson values.

The closer the two histograms look to each other, the better the Poisson distribution approximates the Binomial for those values of `n` and `p`.

For example, calling `plot_comparison(n=10, p=0.5)` shows the comparison for 10 trials with a 50% success rate. You'll use this function in both Question 5.1 and Question 5.2.

```python
# Don't change this cell!
def plot_comparison(n, p, num_data=1000):
    """Plot histograms of Binomial(n, p) and Poisson(n * p) samples."""
    np.random.seed(42)
    lam = n * p

    poisson_data = np.random.poisson(lam, num_data)
    binom_data = np.random.binomial(n, p, num_data)

    plt.figure(figsize=(10, 5))
    plt.hist(binom_data, bins=30, alpha=0.6, color='green',
             label='Binomial (count)', edgecolor='grey')
    plt.hist(poisson_data, bins=30, alpha=0.6, color='red',
             label='Poisson (count)', edgecolor='grey')
    plt.xlabel("value")
    plt.ylabel("Count")
    plt.legend()
    plt.show()
```

```python
# Compares a Binomial and Poisson Distribution where n=10 and p=0.5
# Notice how the height and the spread of the distributions look fairly different
plot_comparison(10, 0.5)
```
